# Airline Passenger Satisfaction – Data Cleaning

This notebook prepares the raw airline passenger satisfaction dataset for analysis by handling missing values, duplicates, inconsistent column names, and unnecessary columns.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("../data/raw/airline_passenger_satisfaction.csv")
```

```python
print(f"Initial rows: {df.shape[0]}")
print(f"Initial columns: {df.shape[1]}")
```

```python
df.columns=(df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
 )
```

```python
df.columns.tolist()
```

```python
missing_values= df.isnull().sum()
missing_values=[missing_values > 0]
print(missing_values)
```

```python
df = df.drop_duplicates()
df.duplicated().sum(
)
```

```python
df.dtypes
```

```python
df[[
    "departure_delay",
    "arrival_delay"
]].isnull().sum()
```

```python
df["arrival_delay"]=(
    df["arrival_delay"]
    .fillna(df["arrival_delay"].median())
)
```

```python
if "id" in df.columns:
    df=df.drop(columns=["id"])
```

```python
df.columns.tolist()
```

```python
print("Gender:")
print(df["gender"].unique())

print("\nCustomer Type:")
print(df["customer_type"].unique())

print("\nTravel Type:")
print(df["type_of_travel"].unique())

print("\nClass:")
print(df["class"].unique())

print("\nSatisfaction:")
print(df["satisfaction"].unique())
```

```python
categorical_columns = [
    "gender",
    "customer_type",
    "type_of_travel",
    "class",
    "satisfaction"
]

for column in categorical_columns:
    df[column] = df[column].str.strip()
```

```python
df[
    [
        "age",
        "flight_distance",
        "departure_delay",
        "arrival_delay"
    ]
].describe()
```

```python
print("Negative ages:", (df["age"] < 0).sum())
print("Negative flight distances:", (df["flight_distance"] < 0).sum())
print(
    "Negative departure delays:",
    (df["departure_delay"] < 0).sum()
)
print(
    "Negative arrival delays:",
    (df["arrival_delay"] < 0).sum()
)
```

```python
df.to_csv(
    "../data/processed/cleaned_airline_data.csv",
    index=False
)
```

```python
df = df.rename(columns={
    "Departure Delay in Minutes": "departure_delay_in_minutes",
    "Arrival Delay in Minutes": "arrival_delay_in_minutes"
})
```
