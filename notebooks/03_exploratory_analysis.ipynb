# Airline Passenger Satisfaction – Exploratory Data Analysis

This notebook explores passenger demographics, travel characteristics, service ratings, delays, and satisfaction patterns to identify key business insights.

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("../data/processed/cleaned_airline_data.csv")
```

```python
df.describe().T
```

```python
print("Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nTotal Records:")
print(len(df))
```

```python
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

df[numerical_columns].describe().T
```

```python
categorical_columns = df.select_dtypes(include=["object"]).columns

df[categorical_columns].describe().T
```

```python
gender_distribution = df["gender"].value_counts()

print(gender_distribution)
```

```python
customer_distribution = df["customer_type"].value_counts()

print(customer_distribution)
```

```python
print("Average Age:", round(df["age"].mean(), 2))
print("Median Age:", df["age"].median())
print("Youngest Passenger:", df["age"].min())
print("Oldest Passenger:", df["age"].max())
```

```python
age_bins = [0, 25, 35, 45, 55, 65, 100]

age_labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "66+"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=age_bins,
    labels=age_labels
)

age_group_distribution = df["age_group"].value_counts().sort_index()

print(age_group_distribution)
```

```python
travel_type_distribution = df["type_of_travel"].value_counts()

print(travel_type_distribution)
```

```python
class_distribution = df["class"].value_counts()

print(class_distribution)
```

```python
print("Average Flight Distance:", round(df["flight_distance"].mean(), 2))
print("Median Flight Distance:", df["flight_distance"].median())
print("Shortest Flight:", df["flight_distance"].min())
print("Longest Flight:", df["flight_distance"].max())
```

```python
distance_by_class = df.groupby("class")["flight_distance"].agg(
    ["count", "mean", "median", "min", "max"]
)

distance_by_class.round(2)
```

```python
df["departure_delay"].describe()
```

```python
df["arrival_delay"].describe()
```

```python
departure_delayed = (df["departure_delay"] > 0).sum()
arrival_delayed = (df["arrival_delay"] > 0).sum()

print("Passengers with Departure Delays:", departure_delayed)
print("Passengers with Arrival Delays:", arrival_delayed)

print(
    "Departure Delay Percentage:",
    round((departure_delayed / len(df)) * 100, 2),
    "%"
)

print(
    "Arrival Delay Percentage:",
    round((arrival_delayed / len(df)) * 100, 2),
    "%"
)
```

```python
delay_by_satisfaction = df.groupby("satisfaction")[
    ["departure_delay", "arrival_delay"]
].mean()

delay_by_satisfaction.round(2)
```

```python
satisfaction_distribution = df["satisfaction"].value_counts()

print(satisfaction_distribution)
```

```python
satisfaction_percentage = (
    df["satisfaction"].value_counts(normalize=True) * 100
)

satisfaction_percentage.round(2)
```

```python
satisfaction_by_gender = pd.crosstab(
    df["gender"],
    df["satisfaction"],
    normalize="index"
) * 100

satisfaction_by_gender.round(2)
```

```python
satisfaction_by_customer = pd.crosstab(
    df["customer_type"],
    df["satisfaction"],
    normalize="index"
) * 100

satisfaction_by_customer.round(2)
```

```python
satisfaction_by_class = pd.crosstab(
    df["class"],
    df["satisfaction"],
    normalize="index"
) * 100

satisfaction_by_class.round(2)
```
